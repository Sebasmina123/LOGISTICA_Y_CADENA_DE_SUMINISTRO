### Código hecho para traer los tipos exactos de datos y nombres de cada tabla hecha.

````python
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()

esquemas = ["bronze", "silver", "gold"]

for esquema in esquemas:
    print("\n")
    print("##### ", "="*100)
    print(f"##### CAPA: {esquema.upper()}")
    print("##### ", "="*100)

    tablas = spark.sql(f"SHOW TABLES IN {esquema}")

    for fila in tablas.collect():

        nombre_tabla = f"{esquema}.{fila.tableName}"
        print("\n")
        print("##### ", "-"*100)
        print(f"TABLA: {nombre_tabla}")
        print("##### ","-"*100)

        df = spark.table(nombre_tabla)
        print("```text")
        df.printSchema()
        print("```")
````
