#### En este primer bloque tenemos una función para crear un logger que sirve para el registro de cada ingesta q se haga mediante este script

```python
import uuid
import time
from datetime import datetime
from pyspark.sql.functions import current_timestamp, lit, year, month, dayofmonth
from pyspark.sql.types import StructType, StructField, StringType, TimestampType, LongType, DoubleType

#----- creador de logger ------- #
def registrar_log (nombre_proceso, inicio_tiempo, df_final):
        fin_tiempo = time.time()
        duracion = round(fin_tiempo - inicio_tiempo,2)

        #contar registros
        total_registros = df_final.count()
        
        # esquema de la tabla de logs
        esquema = StructType([
        StructField("Fecha_Ejecucion", TimestampType(), True),
        StructField("Proceso", StringType(), True),
        StructField("Registros_Procesados", LongType(), True),
        StructField("Duracion_Segundos", DoubleType(), True)
        ])

        datos = [(datetime.now(), nombre_proceso, total_registros, duracion)]
        df_log = spark.createDataFrame(datos, esquema)

        df_log.write.format("delta").mode("append").saveAsTable("logging.control_logging")
        print(f"log registrado: procesadas {total_registros} filas, en {duracion}segundos")


```

### Iniciamos el proceso de agregación de metadatos y particiones para cada tabla

```python
#------------ proceso --------#

inicio = time.time()
sistema_origen = "delta_cal_destinatarios" 
id_lote_unico = str(uuid.uuid4())
t = "ingesta_delta.delta_cal_destinatarios"
to = "cal_destinatarios"
fecha_actual = current_timestamp()

try:
        df = spark.read.table(t)
        print(f"leyendo:{t}")

        df_b = (df.withColumn("Marca_tiempo_ingesta", fecha_actual)
                .withColumn("sistema_fuente", lit(sistema_origen))
                .withColumn("id_lote_proc", lit(id_lote_unico))
                .withColumn("year_intake", year(fecha_actual))
                .withColumn("month_intake", month(fecha_actual))
                .withColumn("day_intake", dayofmonth(fecha_actual))
                )

        display(df_b)
        print("guardando...")

        df_b \
        .write \
        .format("delta") \
        .mode("overwrite") \
        .partitionBy("year_intake","month_intake","day_intake") \
        .saveAsTable( f"bronze.bronze_{to}")

        registrar_log(to, inicio, df_b)
        print("exito")

except Exception as e:
        print(f"Fallo. Error en el procesamiento {str(e)}")
```
