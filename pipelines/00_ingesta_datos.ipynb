### INGESTACIÓN Y TRANSFORMACIÓN DE DATOS EN ARCHIVOS TIPO CSV/PARQUET A DELTA TABLES

En este primer notebook hacemos que los archivos que subimos (los archivos parquet y csv) los combirtamos en delta tables para poder darles un mejor manejo durante el proceso de la arquitectura medallion al pasar capa por capa.



### Para cada tabla:
##### se tiene que leer el archivo originario de la carpeta de Files en el lakehouse y luego con Pyspark sobreescribimos y convertimos en formato delta. Posteriormente, guardamos en un nuevo schema creado en la carpeta de Tables (en este caso, ingesta_delta) para que el notebook de bronze pueda tomar las tablas de alli y agregar metadatos.

```python

print("cargando: Files/CAL_DESTINATARIOS.parquet")
df = spark.read.parquet("Files/CAL_DESTINATARIOS.parquet") 

display(df)

df.write \
.mode("overwrite") \
.format("delta") \
.saveAsTable("ingesta_delta.delta_CAL_DESTINATARIOS")
```
